```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris

iris= load_iris()

from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size = 0.3) 
```

# Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
a = LogisticRegression(solver = 'lbfgs', max_iter = 1000)
a.fit(x_train, y_train)
a.score(x_test, y_test)
```

# Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
b = RandomForestClassifier(n_estimators = 30)
b.fit(x_train, y_train)
b.score(x_test, y_test)
```

# KFold cross validation

```python
from sklearn.model_selection import KFold
c = KFold (n_splits = 4)
c
```

```python
for train_index, test_index in c.split(iris.data):
    print(train_index, test_index)
```

```python
def get_score(model, x_train, x_test, y_train, y_test):
    model.fit(x_train, y_train)
    return model.score(x_test, y_test)
```

```python
from sklearn.model_selection import StratifiedKFold
d = StratifiedKFold(n_splits = 4)

score_LR = []
score_RF = []

for train_index, test_index in d.split(iris.data, iris.target):
    x_train, x_test, y_train, y_test = iris.data[train_index], iris.data[test_index], \
                                       iris.target[train_index], iris.target[test_index]
    score_LR.append(get_score(LogisticRegression(solver = 'lbfgs', max_iter = 1000), x_train, x_test, y_train, y_test))
    score_RF.append(get_score(RandomForestClassifier(n_estimators = 30), x_train, x_test, y_train, y_test))
```

```python
score_LR
```

```python
score_RF
```

# cross_val_score function


```python
from sklearn.model_selection import cross_val_score
cross_val_score(LogisticRegression(solver = 'lbfgs', max_iter = 1000), iris.data, iris. target, cv = 4)
```

```python
cross_val_score(RandomForestClassifier(n_estimators = 30), iris.data, iris.target, cv = 4)
```

```python
score_1 = cross_val_score(LogisticRegression(solver = 'lbfgs', max_iter = 1000), iris.data, iris. target, cv = 50)
np.average(score_1)
```

```python
score_2 = cross_val_score(RandomForestClassifier(n_estimators = 30), iris.data, iris.target, cv = 50)
np.average(score_2)
```
