```python
import tiktoken

encoder = tiktoken.get_encoding('cl100k_base')

texts = [
    'hello world'
    'who are you'
    'def fibonacci(n): return n if n<= 1 else fibonacci(n+1) + fibonacci(n-2)'

]

for i in texts:
    tokens = encoder.encode(i)
    print(f'text: {i}')
    print(f'tokens: {tokens}')
    print(f'count: {len(tokens)}')
```

## TEMPERATURE (how long/short it would give the answer)

```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv
load_dotenv()

ques = "give me shares of blue"

for i in [0.0, 0.5, 0.7]:
    llm = ChatGroq(model = "openai/gpt-oss-20b", temperature = i)
    resp = [llm.invoke(ques).content for _ in range(3)]
    print(f'temperature {i}: {resp}')
```

### Max_tokens (as the tokens increases, so does the lenght of the answer)

```python
llm_short_answer = ChatGroq(
    model = "openai/gpt-oss-20b",
    temperature = 0.8,
    max_tokens= 30
)

llm_long_answer = ChatGroq(
    model = "openai/gpt-oss-20b",
    temperature = 0.8,
    max_tokens= 1000
)

Q = 'what is molecular docking'
print ("short:", llm_short_answer.invoke(Q).content)
print ('\nlong:', llm_long_answer.invoke(Q).content)
```

### System prompt or SystemMessage (can be used to give anwer like real AI and can also limit the domain)

```python
from langchain_core.messages import SystemMessage, HumanMessage

# llm = ChatGroq(model = "openai/gpt-oss-20b")

# # without system prompt
# resp_1 = llm.invoke('what is a protein')
# print('without system prompt:')
# print(resp_1.content)
# print('\n', '-'*100, '\n')

# with system prompt (can be used to limit the domain range) and while loop

memory = []

while True:
    try:
        user_input = input('you: ').strip()
        if user_input.lower() in ['q', 'quit', 'exit'] or not user_input:
            break

        messages_crickter = [
        SystemMessage(content ='you are a cricket coach. Use only cricket terminology to answer. Always finish within 2 lines'),
        HumanMessage(content = user_input)
        ]   

        memory.append(HumanMessage(content = user_input))

        resp_2 = llm.invoke(memory)

        print('cricket coach:')
        print(resp_2.content)
        memory.append(SystemMessage(content = resp_2.content))
        print('\n', '-'*100, '\n')

    except (KeyboardInterrupt, EOFError):
        break
```

### invoke - wait for full response (the entire answer will first be finished in the backend and then will be shown to the user)

```python
llm = ChatGroq(model = "openai/gpt-oss-20b")

resp_3 = llm.invoke('count from 1 to 10 slowly')
print('invoke (all at once): ')
print(resp_3.content)
```

### stream - token by token answer (as soon as the answer is approved in the backend, it will be shown to the user)

```python
for i in llm.stream('count from 1 to 20 slowly'):
    print(i.content, end ="", flush = True)
    print()
```

### BATCH - multiple prompts( or SystemMessages) in one call


```python
Ques = [
    'what is molecular docking',
    'what is molecular sceening'
    'what is molecular encoding'
]

resp_4 = llm.batch(Ques)
for q,r in zip(Ques, resp_4):
    print(f'Quesion: {q}')
    print(f'answer: {r.content[:100]}....\n')
```

#### PromptTemplate( or ChatPromptTemplate)  is used change the SystemMessages and HumanMessages without affecting the main upper framework, hence reducing chances of error

```python
from langchain_core.prompts import ChatPromptTemplate

template = ChatPromptTemplate.from_messages([
    ('system' , 'you are an expert in {domain}. Always answer in {language}'),
    ('human', '{quesion}')
])

chain = template | llm

resp_5 = chain.invoke({
    'domain': 'medicinal chemistry',
    'language': 'simple english',
    'quesion': 'what is medicinal chemistry'
})

print(resp_5.content)
```

### Reuse of Template with different inputs for multiple quesions and answers

```python
inputs = [
    {'domain': 'pharmacology', 'language': 'english', 'quesion': 'what is pharmacokinetics and pharmacodynamics'},
    {'domain': 'pharmacognosy', 'language': 'english', 'quesion': 'what is the meaning of crude drugs'},
    {'domain': 'medicinal chemistry', 'language': 'english', 'quesion': 'what is the drug disgning'}
]

for i in inputs :
    resp_6  = chain.invoke(i)
    print(f'domain: {i['domain']}')
    print(f'answer: {resp_6.content[:150]}.....\n')
```
